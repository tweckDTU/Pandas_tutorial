## Cell types: code and Markdown (text)

Jupyter notebooks are separated into cells. A cell can have executable code, like above, or text, like this cell.

The raw text is written using a simple "language" called [Markdown](https://www.markdownguide.org/getting-started/). Special characters allow you to format the raw text, e.g., make it bold, or a title, or a list, etc. [Here is a Markdown cheat sheet](https://www.markdownguide.org/cheat-sheet/).

If you want to edit the raw text, double-click a Markdown (text) cell. When you are done, hit ESC or SHIFT+ENTER to see the formatted text. You can change a cell type from text to code using the 3-dot menu at the top right of the cell.

## Two modes of Jupyter notebooks

As you have seen above, there are two modes in Jupyter notebooks. In Edit mode, for example when you double-click a Markdown cell, you can edit the contents of the cell. In Command mode, you can quickly navigate from cell to cell and execute them.

1. **Edit mode**. 
    * To ENTER Edit mode: click on a code cell or double-click a Markdown cell.  
2. **Command mode**. 
    * To ENTER Command mode: ESC.  
    * See what cell is highlighted with the vertical blue line.  

**IMPORTANT!!!** In Command mode, there are many keyboard shortcuts for adding cells, deleting cells, and more. **Avoid typing in Command mode unless you are using a verified keyboard shortcut!**. You might do something unexpected.

## Cell outputs

Jupyter notebooks will **automatically print** the last line in the cell, if it doesn't contain an equals sign. No other lines are printed.

For example, compare this...

```python
x = ['apple', 'cat', 'dog']  # define a list
first_element = x[0]  # let's take the first element
first_element  # jupyter notebook will print this
```

...to this:

```python
x = ['apple', 'cat', 'dog']  # define a list
first_element = x[0]  # let's take the first element (nothing is printed -- has equal sign)
```

As we saw earlier, you can clear the output of the cell by clicking the 3 dots to the left of the output. You can also hit "Clear All Outputs" on the top bar to delete all outputs in a notebook.

## Quick reference for what we covered here

**NOTE** that the keyboard shortcuts may be different depending on your computer.

1. Two modes in Jupyter notebook:  
    * Edit. Interact with the content within a cell.  
    * Command. Navigate between cells and execute code. 
    * To switch to Command: ESC.  
    * To switch to Edit: click a code cell or double-click a Markdown cell.  
2. Execute code in a cell:  
    * Click on the cell, then click the arrow to the left of the cell.  
    * CTRL+Enter to execute but stay in the same cell.  
    * SHIFT+ENTER to execute but go to the next cell.
3. Clear the output of a cell:  
    * Click the 3 dots to the left of the output, then click "Clear cell output".  
    * Click on the output, then ALT+DELETE  
4. Clear all cell outputs:  
    * Button "Clear All Outputs" at top of notebook.  
5. Edit text in a markdown cell:  
    * Double-click the cell to enter Edit mode.  
6. Change a cell type:  
    * Upper-right corner of the cell, 3-dots, click "Change cell to..."
