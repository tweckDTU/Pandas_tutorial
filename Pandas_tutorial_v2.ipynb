### Værkstedernes økonomi

I følgende del vil vi kigge på værkstedernes økonomi, deres indkomst og udgifter og hvordan en elektrificering vil påvirke det.

Inden vi kan bruge `pandas`, skal vi første importere `pandas` pakken.

Nu har vi importeret pandas og vi kan gå i gang med at bruge funktioner og variable typer som er defineret i pandas pakken.

I pandas arbejder vi med objekter som hedder DataFrame og DataFrames er 2-dimensionale tabelarisk data strukturer eller simpelthen tabeller med kolonner og rækker.

Der er mange forskellige måder at lave en DataFrame. Her i tutorialen vil vi bruge lister og senere csv-filer.

I følgende vil vi lave en DataFrame fra en liste, som indeholder indkomst fra de forskellige værksteder for situation før og efter elektrificering.

`pandas.DataFrame()` er en funktion som laver en ny DataFrame. Input til fuktionen er:
- `data`: Data som skal lagres i DataFramet. I vores tilfælde indkomst i DKK. 
- `index`: Liste som indeholde rækkenavne. Her værkstedsnavne som vi har gemt i `wrk_shp_complete`
- `columns`: Liste som indeholder kolonnenavne. Her navne fra de to scenarier som vi vil undersøge (2017 og fremtids elektrificeret scenario)

```python
df_income = pandas.DataFrame(data=[[10000,13000],[7000,10000],[20000,21000],[5000,5500],[0,5000],[0,15000],[0,0]],index=wrk_shp_complete,columns=[2017,'scenario'])
```

Nu har vi lavet vores DataFrame. Lad os se hvordan den ser ud.

```python
print(df_income)
```

Vi kan nu printe række og kolone navne med attributerne `.index` og `.column`.

```python
print('Index: ', df_income.index)

print('Columns: ', df_income.columns)
```

Begge funktioner retunerer variable af typen `Index`.

En fordel ved at arbejde med DataFrames er at man nemt kan udtrække et udvalg af dataene.

Fx hvis vi vil udtrække data fra scenario kolonnen for følgende værksteder: ElectronicsWorkshop, BeautySalon, Lighting
kan vi gøre det med DataFrame.loc[]


```python
print(df_income.loc[['ElectronicsWorkshop','BeautySalon','Lighting'],'scenario'])
```
